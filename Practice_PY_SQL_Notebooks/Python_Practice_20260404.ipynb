```python

x = [1,'data',True, 2.2]
x.append(3)
x
```

```python
# Creating a list
numbers = [10, 20, 30, 40]

# Adding an element
numbers.append(50)

# Inserting an element at a specific position
numbers.insert(1, 15)

# Removing an element
numbers.remove(30)

# Sorting the list
numbers.sort()

# Finding the length of the list
length = len(numbers)

# Printing results
print("Updated List:", numbers)
print("Length of List:", length)
```

Explanation
numbers = [10, 20, 30, 40]
→ Creates a list, which is an ordered and mutable collection.
append(50)
→ Adds 50 to the end of the list.
insert(1, 15)
→ Inserts 15 at index 1 (second position).
remove(30)
→ Removes the value 30 from the list.
sort()
→ Arranges the list in ascending order.
len(numbers)
→ Returns the total number of elements in the list.

Tuple

```python
# Tuple example (immutable)
my_tuple = (10, 20, 30, 20)

# Accessing elements
print("Tuple:", my_tuple)
print("First element:", my_tuple[0])

# Set example (unordered, unique elements)
my_set = {10, 20, 30, 20}

# Adding an element
my_set.add(40)

# Removing an element
my_set.remove(20)

# Printing results
print("Set:", my_set)
```

🔸 Tuple
my_tuple = (10, 20, 30, 20)
→ A tuple is an ordered and immutable collection (cannot be changed).
It allows duplicate values (notice 20 appears twice).
You can access elements using an index like my_tuple[0].

🔸 Set
my_set = {10, 20, 30, 20}
→ A set is unordered and stores only unique values.
Duplicate values are automatically removed → {10, 20, 30}
add(40) → Adds a new element
remove(20) → Removes an element

```python
x = {
    '1': "Avic",
    '2': 'Pooja',
    '3': 'Ravi',
    '4': 'Malkeet',
    '5': 'Mahesh',
    '6': 'Jairaj'
}

print(x)
```

```python
# Access specific value
print(x['1'])          # Avic

# Using get()
print(x.get('2'))      # Pooja

# Get all keys
print(x.keys())

# Get all values
print(x.values())

# Get key-value pairs
print(x.items())
```

🔍 Explanation
x['1'] → Direct access using key
get('2') → Safe access (no error if key missing)
keys() → returns all keys
values() → returns all values
items() → returns (key, value) pairs

```python
x = {
    "batch_1": {
        '1': "Avic",
        '2': "Pooja",
        '3': "Ravi"
    },
    "batch_2": {
        '4': "Malkeet",
        '5': "Mahesh",
        '6': "Jairaj"
    }
}

print(x)
```

```python
# Access batch
print(x["batch_1"])

# Access specific value inside batch
print(x["batch_1"]['1'])   # Avic

# Using get()
print(x.get("batch_2").get('4'))   # Malkeet

# Get all batch names
print(x.keys())

# Get values inside batch_1
print(x["batch_1"].values())
```

🔍 Explanation
x["batch_1"] → accesses full batch
x["batch_1"]['1'] → accesses specific student
get() → safer way to avoid errors
You can chain access: x["batch_2"]['4']
