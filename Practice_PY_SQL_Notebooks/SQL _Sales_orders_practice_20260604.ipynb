## The SQL SELECT Statement

```python
%sql
SELECT * 
FROM batch1.data.sales_orders;
```

```python
%sql
SELECT *
FROM batch1.data.sales_orders
LIMIT 10;
```

SELECT is used to retrieve specific columns or full data from a table.

```python
%sql
SELECT customer_id, customer_name, order_number
FROM batch1.data.sales_orders;
```

```python
%sql
SELECT customer_id, order_number, qty
FROM batch1.data.sales_orders
LIMIT 5;
```

## SQL SELECT DISTINCT Statement


SELECT DISTINCT is used to return only unique (non-duplicate) values from one or more columns.

```python
%sql
SELECT DISTINCT customer_id
FROM batch1.data.sales_orders;
```

```python
%sql
-- Unique customers (ID + name)
SELECT DISTINCT customer_id, customer_name
FROM batch1.data.sales_orders;
```

```python
%sql
-- Unique product + price combinations
SELECT DISTINCT product_id, price
FROM batch1.data.sales_orders;
```

DISTINCT works on combination of columns
If you use multiple columns → uniqueness is based on all of them together

## SQL WHERE Clause


The WHERE clause is used to filter records and

 is used to extract only those records that fulfill a specific condition.

```python
%sql
-- 1. Get all orders with price, qty, order_id
SELECT order_number, qty, price
FROM batch1.data.sales_orders LIMIT 1000;

-- 2. Orders where qty > 2
SELECT order_number, qty, price
FROM batch1.data.sales_orders
WHERE qty > 2;

-- 3. Orders where price > 100
SELECT order_number, qty, price
FROM batch1.data.sales_orders
WHERE price > 100;
```

## Order By

The ORDER BY keyword is used to sort the result-set in ascending or descending order.

The ORDER BY keyword sorts the result-set in ascending order (ASC) by default.

```python
%sql
-- Sort orders by price (highest first)
SELECT order_number, qty, price
FROM batch1.data.sales_orders
ORDER BY price DESC;

-- Sort by qty descending
SELECT order_number, qty, price
FROM batch1.data.sales_orders
ORDER BY qty DESC;
```

```python
%sql
SELECT *
FROM batch1.data.sales_orders
ORDER BY order_datetime;
```
